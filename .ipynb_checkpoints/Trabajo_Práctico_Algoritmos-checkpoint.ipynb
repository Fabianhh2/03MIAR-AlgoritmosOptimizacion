## Problema: 1. Sesiones de doblaje 


### Descripción del problema:

Se precisa coordinar el **doblaje de una película**. Los actores del doblaje deben coincidir en las tomas en las que sus personajes aparecen juntos en las diferentes tomas. Los actores de doblaje cobran todos la misma cantidad por cada día que deben desplazarse hasta el estudio de grabación independientemente del número de tomas que se graben. No es posible grabar más de 6 tomas por día. El objetivo es planificar las sesiones por día de manera que **el gasto** por los servicios de los actores de doblaje **sea el menor posible**. Los datos son:
* Número de actores: 10
* Número de tomas : 30                                        

```python
# Librerías utilizadas
import numpy as np
import pandas as pd
from typing import List, Set, Tuple, Union, Optional, Literal, FrozenSet
import pickle
from collections import Counter
import random
import itertools
```

# Modelo

## 1.Cómo representar las soluciones
Después de pensar en múltiples opciones para representar la solución, se ha llegado a la conclusión de que una forma adecuada es mediante una estructura tipo Vector de Asignación. Para este problema, la solución que devolverá el algoritmo principal será un vector unidimensional de longitud fija $N=30$.

$$
S = [d_0,d_1,\dots,d_{29}],
$$

donde:
* La posición $i$ del vector representa la **Toma** $i$ (de 0 a 29, 30 tomas en total).
* El valor $S[i]$ representa el **Día** asignado a esa toma.

Se ha decidido expresar de esta forma el espacio de soluciones ya que en el problema hay valores como el `número de días` o `número de tomas por día` que no se conocen y que no necesariamente tienen que ser fijos. No obstante, sí conocemos que el problema tiene 30 tomas y que cada toma se va a a realizar en algún día, por lo que esta propuesta siempre va a devolver soluciones interpretables.

## 2. Función objetivo

El objetivo final del problema es minimizar el **Costo total de de los actores**. Por los datos del problema, sabemos que todos los actores cobran solo por desplazarse al set de grabación, este coste lo definiremos como unidad, y una vez allí pueden grabar hasta 6 escenas. Matemáticamente, esto se representa de la siguiente forma:

### Planteamiento matemático

Sean $d_{max}$ los días totales de grabación, $i \in \{0,1,\dots,29\}$ el índice de las tomas, $j \in \{0,1,\dots,d_{max}\}$ el índice de los días de grabación y $k \in \{0,1,\dots,9\}$ el índice de los actores. Definimos los siguientes elementos:

1. Los conjuntos asociados a las tomas y a los días:
$$
A_i \subseteq \{0,1,\dots,9\} : \text{Conjunto de actores que participan en la toma } i
$$

$$
B_j = \bigcup_{S[i]=j} A_i : \text{Conjunto de actores que participan en el día } j
$$

2. La función objetivo:
$$
f_j = |B_j| : \text{Coste de grabación del día } j
$$

$$
f = \sum_{j=0}^{d_{max}} f_j = \sum_{j=0}^{d_{max}} |B_j| : \text{Coste de grabación total}
$$

---

De este modo se construye la función objetivo, que se llamará `costo_total`. Esta función calculará el coste de realizar toda la producción tomando como parámetros:

* $S$: La solución del problema.
* **Tomas**: Diccionario que contiene la información sobre los actores que participan en cada toma.

Como se pide en el problema, buscaremos **minimizar la función** todo lo que sea posible, es decir, se trata de un problema:
$$
\begin{array}
\text{(P)} & \text{Min } f(x_{ij}) \\
 &  x_{ij} \in \{0,1\} \\
 & i \in \{ 0,1,\dots,29 \} \\
 & j \in \{ 0,1,\dots,d_{max} \}.
\end{array}
$$

### 3. ¿Cómo implemento las restricciones?

1. Coincidencia de actores: Como se ha explicado, una vez el actor acude al set de grabación su coste ya no cambia, ya grabe 1 o 6 tomas ese día. Esto se puede implementar fácilmente utilizando sets de python y operar de la misma forma que se ha planteado en la construcción matemática de la función objetivo.

2. Cada toma se graba solo un día: A la hora de implementar el algoritmo, es importante que una vez una toma se utilice se descarte o elimine para que no se vuelva a utilizar. Para esto, se utilizará un diccionario donde la clave será el índice de la toma y el valor el set con los actores que participan en aquella toma. Luego, una vez utilizada la toma, se elimina del diccionario.

3. Nº de tomas por día: En un mismo día solo se pueden grabar 6 escenas. Esto se resuelve fácilmente realizando un bucle que finalice cuando la variable asociada a tomas grabadas por día alcance 6.

# Análisis
## 1. Orden de complejidad

Este problema pertenece a la categoría de problemas **NP-Hard**, lo que implica que **no existe un algoritmo conocido que pueda resolverlo en un tiempo polinomial**. Aunque es posible verificar cualquier solución en tiempo polinomial (al revisar que se asignen correctamente los actores a las tomas y que no haya más de 6 tomas por día), encontrar la solución óptima implica explorar un gran número de combinaciones posibles, lo cual crece exponencialmente con el número de tomas.

### Parámetros:
- **N**: Número de tomas = 30
- **D**: Número de días necesarios entre 5 y 30 días

Por tanto, nuestro problema (asignar cada toma a uno de los $D$ días) tiene una complejidad de $ O(D^N) $, lo que significa que la cantidad de combinaciones posibles crece exponencialmente.

---

## 2. Contabilización del espacio de soluciones

Consideremos el caso **idóneo** en el que cada día se graban exactamente 6 tomas. Esto llevaría a la siguiente estimación del número de días necesarios: $ \left\lceil \frac{30}{6} \right\rceil = 5 $ días.

### Cálculo del número de combinaciones posibles:
- Día 1. Seleccionamos 6 tomas de 30: $\binom{30}{6}$ = 593775
- Día 2. Seleccionamos 6 tomas de 24: $\binom{24}{6}$ = 134596
- Día 3. Seleccionamos 6 tomas de 18: $\binom{18}{6}$ = 18564
- Día 4. Seleccionamos 6 tomas de 12: $\binom{12}{6}$ = 924
- Día 5. Seleccionamos 6 tomas de 30: $\binom{6}{6}$ = 1

El producto nos da el total de combinaciones posibles:  $1370874167589326400 \approx 10^{18}$.

Sin embargo, en este total se incluyen combinaciones redundantes, ya que las diferentes asignaciones de las tomas a los días no son necesariamente únicas. Esto se debe a que el **orden de las jornadas de grabación no importa**. Para corregir esto, se debe dividir el número total de combinaciones por el factorial del número de grupos (5!):

$$
\frac{1370874167589326400}{5!} = 114239513965777720 \approx 10^{17}
$$

# Diseño: Técnica voraz con búsqueda de candidatos

## 1. Explicación

Se van a relizar 3 versiones del algoritmo, en los que progresivamente se va a ir añadiendo complejidad y nuevos criterios para mejorar la búsqueda de la solución. No obstante, las 3 versiones comparten una estructura básica:

1. Función(es) auxilar(es): Busca(n) una toma semilla que genera una lista de candidatos y luego seleccionan al mejor según un criterio.
2. Función principal: Construye la solución actualizando los datos y reiniciando la(s) función(es) auxiliar(es).

El enfoque general es el siguiente: *Quiero buscar una toma buena y a partir de ahí ir colocando el resto de tomas.*

## 2. ¿Por qué esta técnica?

El problema presenta un espacio de soluciones muy grande, lo que hace descartar el uso de algoritmos que construyen muchas posibles soluciones como por ejemplo el backtracking. Por esta razón, se ha optado por un algoritmo voraz con construcción de candidatos, ya que se trata de un método eficiente y determinista.

En cada iteración, el algoritmo elige al mejor candidato según una función de selección basada en uno o varios criterios. Luego, a partir de este candidato se eligen las tomas que mejor combinan con este hasta completar el día.

Aunque esta función de selección es más compleja que en los algoritmos voraces clásicos, el esquema sigue siendo voraz, ya que:
1. La solución se construye por etapas.
2. En cada etapa se elige el mejor candidato disponible.
3. No se reconsideran decisiones anteriores.

Esto permite mantener un equilibrio entre eficiencia e interpretabilidad, ya que la lógica voraz viene determinada por los criterios de selección. Sin embargo, la limitación de este algoritmo radica en que no se ha demostrado si tomar la mejor decisión en cada etapa lleva al mínimo global, por lo que se desconoce si la mejor solución obtenida es la mejor posible.

```python
lista_conjuntos = dict(df_actores.apply(lambda row: set(row[row].index), axis=1))
Tomas = {j:{i for i,val in enumerate(matriz_actores[j]) if val} for j in range(len(matriz_actores))}
print(Tomas)
```

## Funciones básicas para trabajar y visualizar las soluciones

```python
# Establecemos la naturaleza de los parámetros
SolutionType = List[int]
NumericType = Union[float,int]
TakesType = List[set[int]]
MAX_TOMAS_POR_DIA = 6

def costo_total(solucion: SolutionType, tomas: TakesType) -> int:
    """
    Cálculo del costo total a la solución de entrada. Los actores de doblaje
    cobran todos la misma cantidad por cada día que deben desplazarse hasta
    el estudio de grabación independientemente del número de tomas que se graben
    Parámetros:
    - Solucion
    - Tomas
    """
    # 1. Crea una lista de conjuntos. Hay tantos conjuntos como días de grabación
    actores_por_dia = [set() for _ in range(max(solucion) + 1)]
    
    # 2. Rellenamos los conjuntos. Así sabemos qué actores van ese día a grabar
    for toma, dia in enumerate(solucion):
        actores_por_dia[dia].update(tomas[toma])
        
    # 3. Sumamos el nº de actores por día por cada día de grabación
    return sum(len(s) for s in actores_por_dia)

def tabla_solucion(solucion: SolutionType, tomas: TakesType) -> None:
    """
    Método que visualiza el resultado del formato Vector de asignación en una
    tabla de pandas.\n
    Código sugerido por Gemini 3.1 Pro con modificaciones hechas por el autor
    Parámetros
    - Solucion
    - Tomas
    """
    num_dias = max(solucion) + 1
    dias_agrupados = [[] for _ in range(num_dias)]
    for toma_id, dia_id in enumerate(solucion):
        dias_agrupados[dia_id].append(toma_id)
    max_slots = 6
    filas = []
    costo_total = 0
    for dia_id, tomas_del_dia in enumerate(dias_agrupados):
        if not tomas_del_dia:
            continue  # saltar días sin tomas
        tomas_del_dia.sort()
        fila = {"Jornada": f"Día {dia_id}"}
        actores_dia = set()
        for slot in range(max_slots):
            col_name = f"Sesión Toma {slot+1}"
            if slot < len(tomas_del_dia):
                toma_id = tomas_del_dia[slot]
                actores = sorted(list(tomas[toma_id]))
                actores_str = ",".join(map(str, actores))
                fila[col_name] = f"Toma {toma_id}, Actores: {actores_str}"
                actores_dia.update(actores)
            else:
                fila[col_name] = ""
        c_dia = len(actores_dia)
        costo_total += c_dia
        fila["Costo"] = c_dia
        filas.append(fila)
        fila_total = {"Jornada": "Total"}
    for slot in range(max_slots):
        fila_total[f"Sesión Toma {slot+1}"] = ""
    fila_total["Costo"] = costo_total
    filas.append(fila_total)
    df = pd.DataFrame(filas)
    df.set_index("Jornada", inplace=True)
    display(df)
```

# 1ª versión del algoritmo

El enfoque voraz que se va a utilizar será el siguiente

1. Para un día, se selecciona una toma con muchos actores.
2. Con esos actores, que ya están en el plató, se hacen el máximo número de escenas (respetando las restricciones).
3. Una vez hecho esto, se guarda la solución y se pasa al siguiente día.

Este enfoque simple nos va a ayudar a entender cómo se comportan las soluciones. 

# Análisis de las funciones

## Complejidad de `encontrar_subconjuntos`

Sea $n$ el número de tomas del diccionario y $k$ el tamaño máximo de los conjuntos.

1. **Búsqueda de subconjuntos**: se recorren $n$ elementos y cada comprobación cuesta $O(k)$  $\Rightarrow O(n \cdot k)$

2. **Ordenación**: en el peor caso se ordenan $n$ elementos  $\Rightarrow O(n \log n)$

3. **Selección de los 6 primeros**: coste constante  $\Rightarrow O(1)$

### Complejidad total

$$
O(n \cdot k + n \log n)
$$

Además, la complejidad espacial es $O(n)$, ya que se almacena una lista auxiliar con hasta $n$ elementos.

## Complejidad de `algoritmo_voraz_v1`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.

- El bucle principal puede ejecutarse hasta $n$ veces.
- En cada iteración, el coste dominante es la llamada a `encontrar_subconjuntos`:
  
  $$O(n \cdot k + n \log n)$$

### Complejidad total

$$O(n^2 k + n^2 \log n)$$

La complejidad espacial es $O(n)$.

## Funciones auxiliares para el algoritmo voraz

```python
def encontrar_subconjuntos(diccionario, conjunto_principal):
    # 1. Buscamos los subconjuntos
    tomas_realizadas = [(clave,conjunto) for clave, conjunto in diccionario.items() if conjunto.issubset(conjunto_principal)]

    # 2. Ordenamos la lista según la longitud de los conjuntos
    tomas_realizadas = sorted(tomas_realizadas,key=lambda item: len(item[1]),reverse=True)
    
    # 3. Nos quedamos solo con los 6 primeros
    tomas_seleccionadas,_ = zip(*tomas_realizadas[:6])
    return tomas_seleccionadas
```

```python
def algoritmo_voraz_v1(tomas:TakesType) -> SolutionType:
    '''
    Realiza un cálculo de las soluciones mediante técnica voraz
    param:
    *tomas: Conjunto de actores que participan en cada escena
    '''
    # 1. Creamos el array de soluciones 
    solucion = np.zeros(len(tomas),dtype=int)
    dia = 0
    tomas = tomas.copy()
    # Bucle principal
    while True: # Se parará cuando la lista se complete
        dia += 1
        # 2. Seleccionamos la toma con más actores
        toma,conjunto = max(tomas.items(),key=lambda item: len(item[1]))
        
        # 3. Buscamos si hay subconjuntos de esa toma
        tomas_dia = list(encontrar_subconjuntos(tomas,conjunto))
        # 4. Actualizamos la solución
        solucion[tomas_dia] = dia
        
        # 5. Eliminamos los elementos del diccionario
        for i in tomas_dia:
            del tomas[i]
        
        # 6. Verificamos que ya no quedan tomas por rodar
        if not tomas:
            return solucion
```

## Solución

```python
sol1 = algoritmo_voraz_v1(Tomas)
tabla_solucion(sol1,Tomas)
```

# 2ª versión del algoritmo

Esta vez, en vez de seleccionar una toma, seleccionaremos 2 tomas cuya unión genere un conjunto lo suficientemente grande como para tener al menos 6 subconjuntos (incluidas las tomas)

1. Para un día, se seleccionan 2 tomas.
2. A partir de la unión, se buscan combinaciones de tomas (subconjuntos de la unión) que permitan cubrir el mayor número de escenas posibles.
3. Se repite este proceso hasta n veces y se selecciona la mejor combinación de tomas según los siguientes ordenados por impotancia: 
    1. Nº de tomas.
    2. coste.
    3. Nº de actores.
4. Se asignan las tomas elegidas al día correspondiente, se eliminan las tomas de la lista de candiatos y se repite el proceso.

Este enfoque introduce una búsqueda más elaborada de candidatos, mejorando la calidad de las soluciones a costa de mayor complejidad.

# Análisis de las funciones

## Complejidad de `seleccion_tomas`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.

- Se recorren todos los pares de tomas:
  
  $$O(n^2)$$

- Para cada par:
  - Operaciones de conjuntos: $O(k)$
  - Llamada a `encontrar_subconjuntos`: $O(n \cdot k + n \log n)$

### Complejidad total

$$
O(n^2 \cdot (n \cdot k + n \log n)) = O(n^3 k + n^3 \log n)
$$

La complejidad espacial es $O(n)$.

## Complejidad de `algoritmo_voraz_v2`

- El bucle principal puede ejecutarse hasta $n$ veces.
- En cada iteración, el coste dominante es la llamada a `seleccion_tomas`:

$$O(n^3 k + n^3 \log n)$$

### Complejidad total

$$
O(n^4 k + n^4 \log n)
$$

La complejidad espacial es $O(n)$.

```python
def costo_grupo_tomas(tomas: TakesType, tomas_seleccionadas) -> int:
    """
    Calcula el coste de grabar un conjunto de tomas en un mismo día.
    El coste es el número de actores distintos que participan.
    """
    actores = set()
    for t in tomas_seleccionadas:
        actores.update(tomas[t])
    return len(actores)

def seleccion_tomas(tomas: TakesType, min_intersec= 2, min_return= 3):
    """
    Busca pares de tomas cuya unión permita cubrir varias tomas como subconjuntos.
    Devuelve la mejor selección encontrada según:
    1) máximo número de tomas cubiertas
    2) mínimo número de actores en la unión
    """
    claves = list(tomas.keys())
    mejor_tomas = None
    mejor_num_tomas = -1
    mejor_union_size = float("inf")
    mejor_score = None

    for idx_i in range(len(claves)):
        i = claves[idx_i]
        conjunto_i = tomas[i]

        for idx_j in range(idx_i + 1, len(claves)):
            j = claves[idx_j]
            conjunto_j = tomas[j]

            # 1. Saltamos casos donde uno contiene al otro
            if conjunto_i.issubset(conjunto_j) or conjunto_j.issubset(conjunto_i):
                continue

            # 2. Intersección mínima para considerar la unión
            n_comunes = len(conjunto_i & conjunto_j)
            if n_comunes < min_intersec:
                continue
            
            # 3. Obtenemos nuevo conjunto y los índices de sus subconjuntos
            nuevo_conjunto = conjunto_i | conjunto_j
            total_tomas = list(encontrar_subconjuntos(tomas, nuevo_conjunto))

            num_tomas = len(total_tomas)
            union_size = len(nuevo_conjunto)

            # Mínimo de tomas para aceptación
            if num_tomas < min_return:
                continue

            # Coste real del grupo
            costo = costo_grupo_tomas(tomas, total_tomas)

            # Número de actores distintos del grupo
            num_actores = costo

            score = (len(total_tomas), -costo, -num_actores)

            if mejor_score is None or score > mejor_score:
                mejor_score = score
                mejor_tomas = total_tomas

    return mejor_tomas
```

```python
def algoritmo_voraz_v2(tomas:TakesType) -> SolutionType:
    '''
    Realiza un cálculo de las soluciones mediante técnica voraz
    Parámetro:
    - tomas: Conjunto de actores que participan en cada escena
    '''
    # 1. Creamos el array de soluciones 
    solucion = np.zeros(len(tomas),dtype=int)
    dia = 0
    tomas = tomas.copy()
    # Bucle principal
    while True: # Se parará cuando la lista se complete
        
        dia += 1
        
        # Toma con más actores
        toma,conjunto = max(tomas.items(),key=lambda item: len(item[1]))
        
        # Función de selección
        seleccion = seleccion_tomas(tomas)

        if seleccion is None:
            tomas_dia = [toma] # Hacemos solo una toma este día y seguimos
        else:
            tomas_dia = list(seleccion)
                
        # 4. Actualizamos la solución
        solucion[tomas_dia] = dia
        
        # 5. Eliminamos las tomas ya asignadas
        for i in tomas_dia:
            del tomas[i]
        
        # 6. Verificamos que ya no quedan tomas por rodar
        if not tomas:
            return solucion
```

## Solución

```python
sol2 = algoritmo_voraz_v2(Tomas)
tabla_solucion(sol2,Tomas)
```

### Comentarios de esta 2ª versión
El algoritmo funciona bastante mejor que la primera versión. Haber incluido varios criterios para escoger una combinación de tomas hace que en general los días como mínimo estén completos, al menos todos a excepción del último, que se ha quedado en una única toma.

Por la forma de la solución, parece ser que la solución óptima estará distribuida en 5 días, por lo que en la siguiente versión se supondrá que el número de tomas por día será 6.

Además, se tratará de no seguir aumentando la complejidad, ya que en esta 2º versión por la existencia de bucles dobles se ha aumentado mucho la complejidad.

# 3ª versión del algoritmo

### Explicación de las funciones de selección

En esta versión definitiva, se va a volver a elegir una sola toma como semilla, sin embargo esta vez se seleccionará según su rareza, cualidad que vendrá determinada por los siguientes criterios:
1. Menor frecuencia mínima entre sus actores
2. Mayor número de actores en la toma
3. Mayor suma de rarezas (1 / frecuencia)

De este modo, forzamos al algoritmo a construir el día adaptándose a las tomas más difíciles de combinar. Además, para seleccionar el resto de tomas, se iterará sobre todo el conjunto de candidatos para buscar el mejor candidato según la siguiente función, donde $A$ es el conjunto de actores que hasta el momento participan ese día y $A_i$ el conjunto de los actores de la toma $i$:

$$
f(A_i) = \operatorname{Card}\left( A \setminus A_i \right) - k_1 \cdot \operatorname{Rareza}\left( A_i \right) - \frac{\operatorname{Card}\left(A \cup A_i \right)}{k_2} + \frac{\operatorname{Card}\left(A\right)}{k_3}
$$

Además, si $\min_{i\in \{1,2,\dots,n \}}(f(A_i)) > k_4$, se pasará al siguiente día. El objetivo de la función es el siguiente:

* Calcular el incremento de actores al añadir esta toma.
* Recompensar a la toma por su Rareza
* Recompensar a la toma por el número de actores que ya están en el set.
* Penalizar a la toma si el día ya tiene muchos actores

Los valores de los parámetros $k_1,k_2,k_3,k_4$ se determinarán haciendo un pequeño **GridSearch**.

### Pasos de la función

Una vez definidas las dos funciones de selección, pasamos a describir los pasos del algoritmo:

1. Se calcula la frecuencia de aparición de cada actor en las tomas que quedan por asignar.
2. Se elige una toma semilla priorizando actores raros.
3. A partir de esa semilla, se construye el día de grabación añadiendo tomas sucesivamente con la particularidad de que para añadir una toma se iteran todas las tomas y se elige aquella que minimice la función $f$ .
5. Se asignan las tomas al día correspondiente y se repite el proceso hasta completar la solución.

En resumen, este enfoque mantiene la construcción voraz por etapas, pero la función de selección está bastante más elaborada, basada en la rareza de actores, la compatibilidad entre tomas y el control del tamaño del grupo diario.

# Análisis de las funciones

## Complejidad de `calcular_frecuencia_actores`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.

- Se recorren todas las tomas. $O(n)$
- Para cada toma, se recorren sus actores. $O(k)$

### Complejidad total

$$
O(n \cdot k)
$$

La complejidad espacial es $O(n \cdot k)$ en el peor caso, ya que el diccionario de frecuencias puede almacenar todos los actores distintos.

---

## Complejidad de `seleccion_rareza`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.


- Llamada a `calcular_frecuencia_actores`: $O(n \cdot k)$

- Después, se recorren las $n$ tomas.
- En cada toma se calculan listas, mínimo, suma y longitud sobre hasta $k$ actores: $O(k)$

### Complejidad total

$$
O(n \cdot k)
$$

La complejidad espacial es $O(n \cdot k)$.

---

## Complejidad de `costo_grupo_tomas`

Sea $m$ el número de tomas seleccionadas.

- Se recorren las $m$ tomas seleccionadas.
- En cada una, se actualiza un conjunto con hasta $k$ actores.

### Complejidad total

$$
O(m \cdot k)
$$

Como en este caso $m \leq 6$, puede considerarse:

$$
O(k)
$$

La complejidad espacial es $O(k)$.

---

## Complejidad de `seleccion_tomas_v2`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.


1. **Copia del diccionario**:$O(n)$

2. **Selección de semilla** mediante `seleccion_rareza`:$O(n \cdot k)$

3. **Bucle de construcción del día**:

- Como `max_tomas_dia = 6`, el `while` se ejecuta como mucho 5 veces adicionales tras elegir la semilla, luego es constante.
- En cada iteración se recorren hasta $n$ tomas restantes.
- Para cada toma se realizan operaciones sobre conjuntos y listas de tamaño máximo $k$:$O(k)$

- Por tanto, cada iteración cuesta:$O(n \cdot k)$

- Y como el número de iteraciones es constante, el coste total del bucle sigue siendo:$O(n \cdot k)$

### Complejidad total

$$
O(n \cdot k)
$$

La complejidad espacial es $O(n \cdot k)$.

---

## Complejidad de `algoritmo_voraz_v3`

Sea $n$ el número de tomas y $k$ el tamaño máximo de los conjuntos.

- El bucle principal puede ejecutarse hasta $n$ veces.
- En cada iteración:
  - La selección de la toma de respaldo con `max(...)` cuesta $O(n)$.
  - La llamada a `seleccion_tomas_v2` cuesta: $O(n \cdot k)$

- Por tanto, el coste dominante por iteración es:$O(n \cdot k)$

### Complejidad total

$$
O(n^2 \cdot k)
$$

La complejidad espacial es $O(n \cdot k)$.

```python
def calcular_frecuencia_actores(tomas: TakesType) -> dict[int, int]:
    """
    Calcula cuántas veces aparece cada actor en las tomas restantes.
    
    Parámetros:
    - tomas: diccionario {id_toma: conjunto_de_actores}
    
    Devuelve:
    - diccionario {actor: frecuencia}
    """
    frecuencia = {}

    for actores_toma in tomas.values():
        for actor in actores_toma:
            # Calcula el valor de la clave. Si no tiene, devuelve 0
            frecuencia[actor] = frecuencia.get(actor, 0) + 1

    return frecuencia

def seleccion_rareza(tomas: TakesType) -> int:
    """
    Selecciona la toma semilla en función de la rareza de sus actores y la compatibilidad.

    Criterio lexicográfico:
    1. Menor frecuencia mínima entre sus actores
    2. Mayor suma de rarezas (1 / frecuencia)
    3. Mayor número de actores en la toma

    Parámetros:
    - tomas: diccionario {id_toma: conjunto_de_actores}

    Devuelve:
    - id de la toma elegida como semilla
    """
    frecuencia_actores = calcular_frecuencia_actores(tomas)

    mejor_toma = None
    mejor_score = None

    for toma_id, actores_toma in tomas.items():
        
        # 1. Calculamos la frecuencia de los actores de esta toma
        frecuencias_toma = [frecuencia_actores[actor] for actor in actores_toma]

        # 2. Calculamos la frecuencia mínima de los actores
        frecuencia_minima = min(frecuencias_toma)

        # 3. Calculamos la suma de rareza
        suma_rareza = sum(1 / f for f in frecuencias_toma)

        # 4. Evaluamos el número de actores
        num_actores = len(actores_toma)

        # Cálculo lexicográfico para comparar tomas
        score = (-frecuencia_minima, -num_actores,suma_rareza)

        if mejor_score is None or score > mejor_score:
            mejor_score = score
            mejor_toma = toma_id

    return mejor_toma,frecuencia_actores
```

```python
def costo_grupo_tomas(tomas: TakesType, tomas_seleccionadas) -> int:
    """
    Calcula el coste de grabar un conjunto de tomas en un mismo día.
    El coste es el número de actores distintos que participan.
    """
    actores = set()
    for t in tomas_seleccionadas:
        actores.update(tomas[t])
    return len(actores)

def seleccion_tomas_v2(tomas: TakesType,k1:int,k2:int,k3:int,k4:int,max_tomas_dia: int = 6) -> list[int]:
    """
    Construye un día de grabación a partir de una semilla fija, añadiendo tomas 
    para minimizar el incremento de coste en el día, priorizando la compatibilidad
    con el grupo actual y la rareza de los actores.
    
    Parámetros:
    - tomas: diccionario {id_toma: conjunto_de_actores}
    - max_tomas_dia: número máximo de tomas por día (por defecto 6)

    Devuelve:
    - lista con los índices de las tomas seleccionadas para el día
    """
    tomas = tomas.copy()
    
    # Elige la semilla usando seleccion_rareza
    semilla,frecuencia_actores = seleccion_rareza(tomas)
    
    # Inicializamos el día con la semilla
    tomas_dia = [semilla]
    actores_dia = set(tomas[semilla])  # Conjunto de actores que ya están en el día

    # Mientras no tengamos 6 tomas en el día, seguimos añadiendo
    while len(tomas_dia) < max_tomas_dia:
        mejor_candidato = None
        mejor_incremento = float('inf')  # Queremos minimizar el coste de la toma

        # Evaluamos todas las tomas restantes
        for toma_id, actores_toma in tomas.items():
            if toma_id in tomas_dia:  # Ya está asignada, saltarla
                continue
            
            # Método para seleccionar los elementos
            # 1. Calcular el incremento de actores al añadir esta toma
            incremento = len(actores_toma - actores_dia)
            
            # 2. Recompensa por Rareza
            frecuencias_toma = [frecuencia_actores[actor] for actor in actores_toma]
            suma_rareza = sum(1 / f for f in frecuencias_toma)
            incremento -= suma_rareza * k1
                
            # 3. Recompensa por incluir conjuntos grandes
            tamano = len(actores_toma & actores_dia)
            incremento -= tamano / k2
            
            # 4. Penalización por tener un día con muchos actores
            penalizacion_dia_sobrepoblado = len(actores_dia) / k3
            incremento += penalizacion_dia_sobrepoblado

            # Si el incremento es menor al mejor encontrado, la elegimos
            if incremento < mejor_incremento:
                mejor_incremento = incremento
                mejor_candidato = toma_id
        # Comprobamos que el mejor_incremento no sea muy alto. En caso contrario, pasamos de día
        if mejor_incremento > k4 :
            break
        
        # Si encontramos un candidato, lo añadimos al día
        if mejor_candidato is not None:
            tomas_dia.append(mejor_candidato)
            actores_dia.update(tomas[mejor_candidato])
            # Eliminamos la toma del diccionario
            del tomas[mejor_candidato]
        else:
            break  # Si no encontramos más candidatos, paramos

    return tomas_dia
```

```python
def algoritmo_voraz_v3(tomas:TakesType,k1:int=1,k2:int=1,k3:int=1,k4:int=1) -> SolutionType:
    '''
    Realiza un cálculo de las soluciones mediante técnica voraz
    Parámetro:
    - tomas: Conjunto de actores que participan en cada escena
    '''
    # 1. Creamos el array de soluciones 
    solucion = np.zeros(len(tomas),dtype=int)
    dia = 0
    tomas = tomas.copy()
    # Bucle principal
    while True: # Se parará cuando la lista se complete
        
        dia += 1
        
        # Criterio por si la seleccion de tomas falla
        toma,conjunto = max(tomas.items(),key=lambda item: len(item[1]))
        
        seleccion = seleccion_tomas_v2(tomas,k1,k2,k3,k4,max_tomas_dia=6)
        
        if seleccion is None:
            tomas_dia = [toma] # Hacemos solo una toma este día y seguimos
        else:
            tomas_dia = list(seleccion)
        # 4. Actualizamos la solución
        solucion[tomas_dia] = dia
        
        # 5. Eliminamos las tomas ya asignadas
        for i in tomas_dia:
            del tomas[i]
        
        # 6. Verificamos que ya no quedan tomas por rodar
        if not tomas:
            return solucion
```

## GridSearch

```python
# Grid_search

k1_values = list(range(10))
k2_values = list(range(1,5))
k3_values = list(range(1,5))
k4_values = list(range(10))



mejor_combinacion = None
mejor_resultado = float('inf')

for p1, p2, p3,p4 in itertools.product(k1_values, k2_values, k3_values,k4_values):
    resultado = algoritmo_voraz_v3(Tomas,p1,p2,p3,p4)
    coste = costo_total(resultado,Tomas)
    if coste < mejor_resultado:
        mejor_resultado = coste
        mejor_combinacion = (p1, p2, p3,p4)

print(f"\nÓptimo: {mejor_combinacion} con {mejor_resultado}")
```

En definitiva, el trabajo muestra cómo, en problemas de este tipo, es más importante diseñar bien el criterio de selección que intentar recorrer todo el espacio de soluciones, ya que esto último resulta inviable.
